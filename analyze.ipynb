### Key Features

- **Sex**: Jenis kelamin individu (Male/Female).
- **Age**: Usia individu dalam tahun.
- **Height**: Tinggi badan individu dalam meter.
- **Weight**: Berat badan individu dalam kilogram.
- **Hypertension**: Status hipertensi (Yes/No).
- **Diabetes**: Status diabetes (Yes/No).
- **BMI**: Body Mass Index, hasil perhitungan berat dan tinggi badan.
- **Level**: Tingkat kondisi tubuh berdasarkan BMI (Underweight, Normal, Overweight, Obese).
- **Fitness Goal**: Tujuan kebugaran (misal: Weight Gain, Weight Loss, Maintain Weight).
- **Fitness Type**: Jenis latihan utama (misal: Muscular Fitness, Cardio, Flexibility).
- **Exercises**: Daftar latihan yang direkomendasikan.
- **Equipment**: Alat yang diperlukan untuk latihan.
- **Diet**: Rekomendasi pola makan, termasuk jenis makanan dan minuman.
- **Recommendation**: Saran tambahan terkait pola hidup sehat, latihan, dan konsultasi profesional.
